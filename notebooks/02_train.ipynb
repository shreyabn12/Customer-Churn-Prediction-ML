```python
import sys
sys.path.append('..')

from src.train import load_and_prepare_data, train_all_models, print_results_summary
```

```python
# Load and split data
X_train, X_test, y_train, y_test = load_and_prepare_data()
```

```python
# Train all 4 models with Optuna + MLflow
# You'll see progress printed for each model
all_results, best_name, best_pipeline = train_all_models(
    X_train, X_test, y_train, y_test
)
```

```python
# Print comparison table
results_df = print_results_summary(all_results)
```

```python
import matplotlib.pyplot as plt

results_df.plot(kind='bar', figsize=(12, 5), colormap='coolwarm')
plt.title('Model Comparison — All Metrics')
plt.xlabel('Model')
plt.ylabel('Score')
plt.xticks(rotation=15)
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig('../reports/figures/model_comparison.png', dpi=150)
plt.show()
```
